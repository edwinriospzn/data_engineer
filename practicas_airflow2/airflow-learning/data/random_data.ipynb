### 02 Generating random orders

```python
# base python 3.13.9 conda environment

print('running...')
```

```python
import csv
import random
from datetime import datetime, timedelta
from pathlib import Path

# Define the data directory
DATA_DIR = Path("02")
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Read customers from existing CSV
customers = []
with open(DATA_DIR / "customers.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        customers.append((int(row["customer_id"]), row["name"], row["city"]))

# Read products from existing CSV
products = []
with open(DATA_DIR / "products.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        products.append((int(row["product_id"]), row["name"], int(row["price"]), row["category"]))

print(f"📖 Read {len(customers)} customers from customers.csv")
print(f"📖 Read {len(products)} products from products.csv")

# Generate 100 orders
orders = []
start_date = datetime.now() - timedelta(days=30)

for order_id in range(100, 200):  # 100 orders (IDs 100-199)
    customer_id = random.randint(1, len(customers))
    product_id = random.randint(1, len(products))
    quantity = random.randint(1, 3)
    price = products[product_id - 1][2]  # Get price from products tuple
    amount = price * quantity
    order_date = start_date + timedelta(days=random.randint(0, 30))
    
    orders.append([order_id, customer_id, product_id, quantity, amount, order_date.strftime("%Y-%m-%d")])

# Write orders.csv
with open(DATA_DIR / "orders.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["order_id", "customer_id", "product_id", "quantity", "amount", "order_date"])
    writer.writerows(orders)

print(f"\n✅ Generated orders.csv in {DATA_DIR}/")
print(f"   - Orders generated: {len(orders)}")
print(f"\n📊 Sample order: {orders[0]}")
print(f"💰 Total revenue: ${sum(order[4] for order in orders):,}")
```

### 04 Generating random orders size data

```python
import csv
import random
from pathlib import Path

# Define data directory
DATA_DIR = Path("04")
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Fixed customers (5 customers)
CUSTOMERS = [
    (1, "Ana", "Barcelona"),
    (2, "Beto", "Madrid"),
    (3, "Carlos", "Valencia"),
    (4, "Diana", "Sevilla"),
    (5, "Elena", "Bilbao"),
]

# Write customers.csv (always the same)
with open(DATA_DIR / "customers.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    writer.writerow(["customer_id", "name", "city"])
    writer.writerows(CUSTOMERS)

# Generate orders - CHANGE THIS NUMBER to test different branches
NUM_ORDERS = 79  # Try: 20 (small), 50 (medium), 100 (large)

# Generate orders
orders = []
statuses = ["Pending", "Shipped", "Delivered", "Cancelled"]

for order_id in range(1, NUM_ORDERS + 1):
    orders.append([
        order_id,
        random.randint(1, 5),      # customer_id
        random.randint(100, 500),   # amount
        random.choice(statuses)     # status
    ])

# Write orders.csv
with open(DATA_DIR / "orders.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["order_id", "customer_id", "amount", "status"])
    writer.writerows(orders)

print(f"✅ Generated {NUM_ORDERS} orders in {DATA_DIR}/orders.csv")
print(f"   Branch: {'Small' if NUM_ORDERS < 30 else 'Medium' if NUM_ORDERS < 80 else 'Large'}")
```

### 05 Data generation sensor

```python
import csv
from pathlib import Path

# Define data directory
DATA_DIR = Path("05")
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Customers
CUSTOMERS = [
    (1, "Ana", "Barcelona"),
    (2, "Beto", "Madrid"),
    (3, "Carlos", "Valencia"),
    (4, "Diana", "Sevilla"),
    (5, "Elena", "Bilbao"),
]

# Products
PRODUCTS = [
    (1, "Laptop Pro", 1200, "Electronics"),
    (2, "Wireless Mouse", 45, "Accessories"),
    (3, "USB-C Hub", 80, "Accessories"),
    (4, "Monitor 27\"", 350, "Electronics"),
    (5, "Mechanical Keyboard", 120, "Accessories"),
]

# Orders
ORDERS = [
    (100, 1, 1, 2, 2400, "2026-07-01"),
    (101, 2, 3, 1, 80, "2026-07-02"),
    (102, 3, 2, 3, 135, "2026-07-03"),
    (103, 4, 4, 1, 350, "2026-07-04"),
    (104, 5, 5, 2, 240, "2026-07-05"),
]

# Write customers.csv
with open(DATA_DIR / "customers.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["customer_id", "name", "city"])
    writer.writerows(CUSTOMERS)

# Write products.csv
with open(DATA_DIR / "products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["product_id", "name", "price", "category"])
    writer.writerows(PRODUCTS)

# Write orders.csv
with open(DATA_DIR / "orders.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["order_id", "customer_id", "product_id", "quantity", "amount", "order_date"])
    writer.writerows(ORDERS)

print(f"✅ Generated files in {DATA_DIR}/")
print(f"   - customers.csv: {len(CUSTOMERS)} customers")
print(f"   - products.csv: {len(PRODUCTS)} products")
print(f"   - orders.csv: {len(ORDERS)} orders")
```
