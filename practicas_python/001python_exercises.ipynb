#### Question #1 — FinOps-style aggregation

```python
records = [
    {"service": "Compute", "cost": 120.50},
    {"service": "Storage", "cost": 50.00},
    {"service": "Compute", "cost": 80.00},
    {"service": "Networking", "cost": 30.00},
    {"service": "Storage", "cost": 20.00},
]
```

```python
# with pandas
def calculate_cost_by_service(records):
    records_df=pd.DataFrame(records)
    result_df=records_df.groupby('service').sum('cost').sort_values('cost', ascending=False)#records_df.head(10)
    result=result_df.to_dict()['cost']
    return result

print(calculate_cost_by_service(records))
```

```python
# No pandas
from collections import defaultdict
def calculate_cost_by_service(records):
    totals = defaultdict(float)
    for r in records:
        totals[r["service"]] += r["cost"]
    result = dict(sorted(totals.items(), key=lambda kv: kv[1], reverse=True))

    return result

print(calculate_cost_by_service(records))
```

```python
# No pandas 2
def calculate_cost_by_service(records):
    result = {}
    for record in records:
        service = record["service"]
        cost = record.get("cost", 0)
        result[service] = result.get(service, 0) + cost
    return result
print(calculate_cost_by_service(records))
```

```python
costs = [
    {"project_id": "P001", "service": "Compute", "cost": 1200},
    {"project_id": "P001", "service": "Storage", "cost": 300},
    {"project_id": "P002", "service": "Compute", "cost": 800},
    {"project_id": "P002", "service": "Database", "cost": 1500},
    {"project_id": "P003", "service": "Storage", "cost": 200},
    {"project_id": "P003", "service": "Compute", "cost": 2500},
    {"project_id": "P004", "service": "Compute", "cost": 400},
]
projects = [
    {"project_id": "P001", "department": "Data", "environment": "Production"},
    {"project_id": "P002", "department": "Finance", "environment": "Production"},
    {"project_id": "P003", "department": "Data", "environment": "Development"},
    {"project_id": "P004", "department": "Marketing", "environment": "Development"},
]
```

```python
def analyze_cloud_costs(costs, projects):
    costs_df=pd.DataFrame(costs)
    projects_df=pd.DataFrame(projects)
    merged_df = costs_df.merge(projects_df, on='project_id', how='left')
    grouped_df=merged_df.groupby(['department','service']).sum('cost').sort_values('cost', ascending=False).reset_index()
    grouped_df['classification'] = pd.cut(
    grouped_df['cost'],
    bins=[-float('inf'), 1000, 2000, float('inf')],
    labels=['Low', 'Medium', 'High']
    )
    return grouped_df.to_dict(orient='records')

print(analyze_cloud_costs(costs, projects))
```
